# Streaming Content Strategy Analysis – Quick EDA
Netflix · Amazon Prime · Disney+

This notebook gives a fast overview of the cleaned dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv("../data/cleaned_streaming_titles.csv")
print(df.shape)
df.head()
```

```python
# Platform & content type mix
print(df["platform"].value_counts())
print("\nContent type:")
print(df["content_type"].value_counts(normalize=True).round(3))
```

```python
pd.crosstab(df["platform"], df["content_type"], normalize="index").round(3)
```

```python
# Top genres overall
df["primary_genre"].value_counts().head(15)
```

```python
# Simple visualization
fig, ax = plt.subplots(figsize=(8, 4))
df.groupby(["platform", "content_type"]).size().unstack().plot(kind="bar", stacked=True, ax=ax)
ax.set_title("Content Mix by Platform")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
